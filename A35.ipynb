# 1. What is a Text-to-Math problem?
## A Text-to-Math problem is a mathematical question written in natural language that needs to be converted into a mathematical expression or reasoning process to find the answer.

# 2. Why are agents useful for math reasoning?
## AI agents are useful because they can understand the question, decide which tool to use, perform calculations, and provide a final answer.This makes the system more reliable and capable of handling different types of mathematical problems.

# 3. Difference between normal LLM response vs agent-based reasoning
## Normal LLM | Agent-based system 
## Generates an answer directly | Decides how to solve the problem 
## Mainly relies on model knowledge/reasoning | Can use external tools 
## May make arithmetic mistakes | Can use a calculator for exact calculations 
## Limited to available model capabilities | Can combine multiple tools 
## Usually one-step response	| Can perform multiple steps/actions

# Task 2: Build Text-to-Math-Agent

## Test with:

```python
from langchain_ollama import ChatOllama
from langchain.agents import create_agent
from langchain_core.tools import tool
import re


# LLM


model = ChatOllama(
    model="gemma3:latest",
    temperature=0
)


# Calculator Tool
@tool
def calculator(expression: str) -> str:
    """Use this tool to calculate arithmetic expressions such as 2000 + 500, 25 * 40, or 100 / 5."""

    try:
        expression = expression.replace("^", "**")

        result = eval(
            expression,
            {"__builtins__": {}},
            {}
        )

        return str(result)

    except Exception as e:
        return f"Calculation error: {e}"


# Agent
agent = create_agent(
    model=model,
    tools=[calculator],
    system_prompt="""
You are a mathematical problem-solving agent.

Your responsibilities:

1. Understand the user's mathematical question.
2. Break word problems into steps.
3. Use the calculator tool for arithmetic calculations.
4. Solve percentage problems.
5. Solve simple algebra problems.
6. Explain the solution clearly.
7. Provide the final answer.
"""
)
```

```python
question = "What is 2000 + 500?"

response = agent.invoke(
    "messages": [
        {
            "role": "user",
            "content": question
        }
    ]
)

print(response["messages"][-1].content)
```

```python
for msg in response["messages"]:
    print("TYPE:", type(msg).__name__)
    print("CONTENT:", msg.content)
    print("TOOL CALLS:", getattr(msg, "tool_calls", None))
    print("-" * 50)
```
